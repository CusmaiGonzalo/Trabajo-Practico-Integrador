```python
import math

class Operaciones:
  def Suma(a,b):
    c = a + b
    return c

  def Resta(a,b):
    c = a - b
    return c

  def Multiplicacion(a,b):
    c = a * b
    return c

  def Division(a,b):
    c = a / b
    return c

  def Potencia(a,b):
    c = math.pow(a,b)
    return c

  def Raiz(a,b):
    c = math.pow(a,1/b)
    return c

estado = int(input('1. Suma\n2. Resta\n3. Multiplicacion\n4. Division\n5. Potencia\n6. Raiz\n'))
Ain = int(input('Ingrese el primer numero: '))
Bin = int(input('Ingrese el segundo numero: '))

if(estado == 1):
  print('El resultado es: ',Operaciones.Suma(Ain,Bin))
elif(estado == 2):
  print('El resultado es: ',Operaciones.Resta(Ain,Bin))
elif(estado == 3):
  print('El resultado es: ',Operaciones.Multiplicacion(Ain,Bin))
elif(estado == 4):
  print('El resultado es: ',Operaciones.Division(Ain,Bin))
elif(estado == 5):
  print('El resultado es: ',Operaciones.Potencia(Ain,Bin))
elif(estado == 6):
  print('El resultado es: ',Operaciones.Raiz(Ain,Bin))
else:
  print('Opcion no valida')
```
